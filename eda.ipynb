```python
import pandas as pd
df = pd.read_csv('titanic.csv')

print(df.head())
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (10,6)
```

```python
print(df.head())
print(df.info())
print(df.describe())
```

```python
plt.figure()
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title("Missing Values Heatmap")
plt.show()
```

```python
plt.figure()
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix")
plt.show()
```

```python
sns.pairplot(df[['Survived','Pclass','Age','Fare']], hue='Survived')
plt.show()
```

```python
plt.figure()
sns.histplot(df['Age'], kde=True, bins=30, color='blue')
plt.title("Age Distribution with KDE")
plt.show()

```

```python
plt.figure()
sns.boxplot(x='Pclass', y='Fare', data=df)
plt.title("Fare vs Passenger Class")
plt.show()
```

```python
plt.figure()
sns.violinplot(x='Survived', y='Age', data=df)
plt.title("Age Distribution by Survival")
plt.show()
```

```python
plt.figure()
sns.countplot(x='Survived', hue='Sex', data=df)
plt.title("Survival Count by Gender")
plt.show()
```

```python
plt.figure()
sns.barplot(x='Pclass', y='Survived', data=df)
plt.title("Survival Rate by Class")
plt.show()
```

```python
g = sns.FacetGrid(df, col="Survived")
g.map(sns.histplot, "Age")
plt.show()
```

```python
pivot = df.pivot_table(values='Survived', index='Pclass', columns='Sex')
sns.heatmap(pivot, annot=True, cmap="YlGnBu")
plt.title("Survival Rate Heatmap")
plt.show()
```

```python
plt.figure()
sns.kdeplot(df[df['Survived']==1]['Fare'], label='Survived', fill=True)
sns.kdeplot(df[df['Survived']==0]['Fare'], label='Not Survived', fill=True)
plt.legend()
plt.title("Fare Distribution by Survival")
plt.show()
```

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

plt.figure()
sns.barplot(x='FamilySize', y='Survived', data=df)
plt.title("Survival vs Family Size")
plt.show()
```

```python
sns.jointplot(x='Age', y='Fare', data=df, kind='scatter')
plt.show()
```
